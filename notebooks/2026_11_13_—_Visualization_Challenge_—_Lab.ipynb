```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BLUE, RUST, GRAY = '#4a5d7e', '#b4553f', '#c9d0d9'
sns.set_theme(style='white')
plt.rcParams['text.parse_math'] = False

CRITERIA = [
    'title states a finding',
    'axes labeled with units',
    'right chart for the question',
    'honest scale',
    'readable',
    'no wasted ink',
]

def score(**passed):
    """score(title=False, axes=True, ...) -> prints a scored critique."""
    total = 0
    for name, ok in zip(CRITERIA, passed.values()):
        print(('PASS' if ok else 'FAIL'), '-', name)
        total += bool(ok)
    print(f'\nscore: {total}/6')
    return total


def finish(ax, title, xlabel=None, ylabel=None, fontsize=12):
    """Same helper as Monday's lecture: title, labels, spines, tight_layout."""
    ax.set_title(title, loc='left', fontsize=fontsize)
    if xlabel is not None:
        ax.set_xlabel(xlabel)
    if ylabel is not None:
        ax.set_ylabel(ylabel)
    for spine in ('top', 'right'):
        ax.spines[spine].set_visible(False)
    ax.figure.tight_layout()


# --- game-day sample (same generator as the Monday lecture) ------------------
rng = np.random.default_rng(2026)
hours = np.arange(9, 22)
build_up = np.exp(-((hours - 14.5) ** 2) / (2 * 2.3 ** 2))
storm = np.exp(-((hours - 17.0) ** 2) / (2 * 1.5 ** 2))
GAMES = {'Sep 5': False, 'Sep 19': False, 'Oct 10': True, 'Nov 14': True}

rows = []
for game, rainy in GAMES.items():
    rain_mm = rainy * (0.4 + 3.0 * storm) * rng.gamma(10, 0.1, len(hours))
    for h, b, rain in zip(hours, build_up, rain_mm):
        n_main = rng.poisson(b * 150 + 12)
        n_rain = rng.poisson(rain * 5 + 0.2)
        cats = np.concatenate([
            rng.choice(['Food', 'Merch', 'Drink'], n_main, p=[.45, .22, .33]),
            np.repeat('RainGear', n_rain),
        ])
        rows += [(game, 'rain' if rainy else 'dry', h, round(rain, 1), c) for c in cats]

df = pd.DataFrame(rows, columns=['game', 'weather', 'hour', 'rain_mm', 'category'])
df['zone'] = rng.choice(['A', 'B', 'C'], len(df), p=[.6, .3, .1])
df['revenue'] = (
    df['category'].map({'Food': 12, 'Merch': 28, 'Drink': 6, 'RainGear': 15})
    * rng.lognormal(-0.18, 0.6, len(df))
).round(2)

print(len(df), 'orders across', df['game'].nunique(), 'games')
df.head()
```

### Chart 1 — revenue by category

Sorted horizontal bar. Axis from zero.

```python
# TODO
fig, ax = plt.subplots(figsize=(7, 3.5))
cat_rev = df.groupby('category')['revenue'].sum().sort_values()
ax.barh(cat_rev.index, cat_rev.values, color=BLUE)
finish(ax, 'Merch and Food Drive the Majority of Total Game-Day Sales', xlabel='total revenue (USD)', ylabel=None)
plt.show()
```

**Caption:** Since merchandise and food bring in the most revenue, there should be more inventory invoiced towards those categories.

### Chart 2 — revenue by hour

Line chart, kickoff marked at 15.5, and the title should state the staffing implication rather than describing the axes.

```python
# TODO
fig, ax = plt.subplots(figsize=(8, 3.5))
hourly_rev = df.groupby('hour')['revenue'].sum()
ax.plot(hourly_rev.index, hourly_rev.values, color=RUST, marker='o', linewidth=2)
ax.axvline(15.5, color='gray', linestyle='--', alpha=0.7)
ax.text(15.7, hourly_rev.max() * 0.9, 'Kickoff (15:30)', color='gray', fontsize=10)
finish(ax, 'Staffing levels must peak by 2:30 PM to handle the heavy pre-kickoff rush', xlabel='hour of day', ylabel='total revenue (USD)')
plt.show()
```

**Caption:** Staff members need to be in by 2:30 or earlier, in order to handle the influx of sales an hour before the game starts.

### Chart 3 — revenue by zone, normalized

Raw revenue by zone mostly measures how big each zone is. Use the capacity figures below to build a second view — revenue per unit of capacity — and show both side by side.

Then say which of the two you would put in front of a manager, and for which question.

```python
zone_capacity = {'A': 12000, 'B': 4000, 'C': 1500}
# TODO: two panels -- raw revenue and revenue per capacity

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))

# raw revenue by zone
zone_rev = df.groupby('zone')['revenue'].sum().sort_index()
ax1.bar(zone_rev.index, zone_rev.values, color=BLUE)
finish(ax1, 'Zone A generates the highest raw revenue due to scale', xlabel='zone', ylabel='total revenue (USD)')

# revenue per unit of capacity
zone_per_cap = zone_rev / pd.Series(zone_capacity)
ax2.bar(zone_per_cap.index, zone_per_cap.values, color=RUST)
finish(ax2, 'Zone C delivers superior efficiency per unit of capacity', xlabel='zone', ylabel='revenue per capacity unit (USD)')

plt.tight_layout()
plt.show()
```

**Caption, and which panel I would present:** In regards to raw revenue, I would present zone A, since it brings in the highest raw revenue. For revenue per capacity, I would present zone C, since it has the highest efficiency and revenue per person despite being the smallest zone.

### Chart 4 — distribution, not just totals

Every chart so far is a total. **TODO:** show the *distribution* of individual order revenue with a histogram. What does the shape tell you that a category total does not?

```python
# TODO
fig, ax = plt.subplots(figsize=(7, 3.5))
sns.histplot(df['revenue'], bins=30, color=BLUE, ax=ax, edgecolor='white')
finish(ax, 'Individual order values are heavily right-skewed, with most under $20', xlabel='order revenue (USD)', ylabel='order count')
plt.show()
```

**Caption:** This shape reveals that most individual orders are small (under $20),so high overall category totals are driven by high transaction volume rather than expensive individual items.

### Chart 5 — your choice

**TODO:** a grouped, stacked, or scatter view that reveals something the first four do not. The finding is the point; the chart type is your call.

If you use a scatter, size or annotate the points so a reader can tell which ones carry enough data to trust.

```python
# TODO
# hourly revenue broken down by weather condition
fig, ax = plt.subplots(figsize=(8, 3.5))

weather_hourly = df.groupby(['hour', 'weather'])['revenue'].sum().unstack(fill_value=0)

ax.plot(weather_hourly.index, weather_hourly['dry'], label='Dry Games', color=BLUE, marker='o')
if 'rain' in weather_hourly.columns:
    ax.plot(weather_hourly.index, weather_hourly['rain'], label='Rain Games', color=RUST, marker='s')

ax.axvline(15.5, color='gray', linestyle='--', alpha=0.7)
ax.legend(frameon=False)
finish(ax, 'Rainy games have higher sales peaks than dry games and shifts sales patterns', xlabel='hour of day', ylabel='total revenue (USD)')
plt.show()
```

**Caption:** Sales peak higher on rainy game days than dry game days, so there needs to be weather checks to adjust inventory.
